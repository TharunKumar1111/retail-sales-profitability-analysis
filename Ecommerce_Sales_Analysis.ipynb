```python
date_map = df3.drop_duplicates('order_id').set_index('order_id')['order_date']

df2['order_date'] = df2['order_id'].map(date_map)
```

```python
df2['year'] = pd.to_datetime(df2['order_date']).dt.year
```

```python
avg_price = df2.groupby(
    ['product_id', 'year']
)['unit_price'].transform('mean')
```

```python
df2['unit_price']=df2['unit_price'].fillna(avg_price)
```
